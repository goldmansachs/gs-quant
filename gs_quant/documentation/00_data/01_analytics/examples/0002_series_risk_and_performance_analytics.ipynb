# Series Risk & Performance AnalyticsCompute risk and performance analytics from a price (or return) series directly, without a Marquee report.Available in ``gs_quant.timeseries``:- **Risk**: ``value_at_risk``, ``conditional_value_at_risk`` (CVaR / expected shortfall), ``drawdown``, ``downside_deviation``, ``tracking_error_of``- **Annualization**: ``annualized_volatility``, ``annualized_return``- **Performance**: ``cumulative_returns``, ``win_rate``, ``omega_ratio``, ``upside_capture``, ``downside_capture``

```python
import pandas as pdfrom gs_quant.timeseries import (    annualized_return,    annualized_volatility,    conditional_value_at_risk,    cumulative_returns,    downside_capture,    downside_deviation,    drawdown,    omega_ratio,    tracking_error_of,    upside_capture,    value_at_risk,    win_rate,)
```

```python
# Sample daily price series (deterministic)prices = pd.Series(    [100.0, 102.0, 99.0, 101.0, 105.0, 103.0, 108.0, 107.0, 110.0, 109.0],    index=pd.date_range('2024-01-01', periods=10, freq='B'),)benchmark = pd.Series(    [100.0, 101.0, 100.0, 102.0, 104.0, 105.0, 106.0, 107.0, 108.0, 109.0],    index=prices.index,)
```

## Value at Risk & Expected Shortfall95% historical VaR is the worst loss expected with 95% confidence; CVaR is the average loss beyond that point.

```python
var_95 = value_at_risk(prices, level=0.95)cvar_95 = conditional_value_at_risk(prices, level=0.95)print(f"95% VaR  = {var_95.iloc[-1]:.2%}")print(f"95% CVaR = {cvar_95.iloc[-1]:.2%}")
```

## Drawdown & AnnualisationRunning drawdown from peak, annualised volatility/return, cumulative returns.

```python
dd = drawdown(prices)print(f"Max drawdown = {dd.min():.2%}")print(f"Annualised vol = {annualized_volatility(prices, window=252).iloc[-1]:.2%}")print(f"Annualised return = {annualized_return(prices, window=252).iloc[-1]:.2%}")print(f"Cumulative return = {(cumulative_returns(prices).iloc[-1] - 1):.2%}")
```

## Downside & Benchmark-Relative MetricsDownside deviation, tracking error, upside/downside capture, win rate, and the Omega ratio.

```python
print(f"Downside deviation = {downside_deviation(prices).iloc[-1]:.2%}")print(f"Tracking error vs benchmark = {tracking_error_of(prices, benchmark).iloc[-1]:.2%}")print(f"Upside capture = {upside_capture(prices, benchmark).iloc[-1]:.2%}")print(f"Downside capture = {downside_capture(prices, benchmark).iloc[-1]:.2%}")print(f"Win rate = {win_rate(prices, w=5).iloc[-1]:.2%}")print(f"Omega ratio = {omega_ratio(prices).iloc[-1]:.2f}")
```
