# Portfolio Construction UtilitiesBuild :class:`PositionSet` objects directly from asset identifiers with different weighting schemes.Available in ``gs_quant.markets``:- ``equal_weight(asset_ids)`` - equally-weighted positions- ``market_cap_weighted(asset_ids, market_caps)`` - market-capitalisation weights- ``from_asset_ids(asset_ids, weights)`` - explicit weights (normalised to sum to 1)

```python
from gs_quant.markets import equal_weight, from_asset_ids, market_cap_weighted
```

```python
# Equal weight across three identifiersequal_ps = equal_weight(['AAPL UW', 'MSFT UW', 'GOOG UW'])print([p.identifier for p in equal_ps.positions])print([p.weight for p in equal_ps.positions])
```

```python
# Market-cap weighted (3:1 ratio)caps = {'AAPL UW': 3000.0, 'MSFT UW': 1000.0}cap_ps = market_cap_weighted(list(caps.keys()), market_caps=caps)print([p.weight for p in cap_ps.positions])
```

```python
# Explicit weights are normalised to sum to 1ps = from_asset_ids(['A', 'B', 'C'], weights=[2, 1, 1])print([p.weight for p in ps.positions])print(f"sum = {sum(p.weight for p in ps.positions)}")
```
